```python
churn_count = df["Churn"].value_counts()

churn_rate = (df["Churn"] == "Yes").mean() * 100

print("Churned Customers:", churn_count["Yes"])
print("Retained Customers:", churn_count["No"])
print("Churn Rate:", round(churn_rate, 2), "%")
```

```python
pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100
```

```python
pd.crosstab(
    df["Contract"],
    df["Churn"]
)
```

```python
bins = [0, 12, 24, 48, 72]
labels = ["0–12 months", "13–24 months", "25–48 months", "49–72 months"]

df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
tenure_churn = pd.crosstab(
    df["TenureGroup"],
    df["Churn"],
    normalize="index"
) * 100

tenure_churn
```

```python
payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

payment_churn.round(2)
```

```python
pd.crosstab(
    df["PaymentMethod"],
    df["Churn"]
)
```

```python
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

internet_churn.round(2)
```

```python
pd.crosstab(
    df["InternetService"],
    df["Churn"]
)
```

```python
df.groupby("Churn")["MonthlyCharges"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)
```

```python
df.groupby("Churn")["TotalCharges"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)
```

```python
senior_churn = pd.crosstab(
    df["SeniorCitizen"],
    df["Churn"],
    normalize="index"
) * 100

senior_churn.round(2)
```

```python
senior_churn.index = ["Non-Senior", "Senior"]

senior_churn.round(2)
```
